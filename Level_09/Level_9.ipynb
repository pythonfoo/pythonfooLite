## Komplexe Daten mit Objekten darstellen

Wir starten damit, aus den einfachen Datentypen komplexere zu machen.

Angenommen, du möchtest dein Adressbuch selbst verwalten. Eine Adresse hat mehrere Komponenten: Name, Vorname, Straße, Postleitzahl, Ort, Land, Telefonnummer, E-Mail. In Python können wir diese Eigenschaften in einer **Klasse** zusammenfassen:

````python
class Adresse:
    name = ""
    vorname = ""
    strasse = ""
    plz = ""
    ort = ""
    land = ""
    telefon = ""
    email = ""
````

Wie du schon aus deinem ersten Programm weißt, werden Klassennamen in Python großgeschrieben (*UpperCamelCase*). Die Eigenschaften einer Klasse werden wie Zuweisungen geschrieben - es sind zunächst **Klassenvariablen**. In unserem Beispiel sind alle Eigenschaften mit leeren Zeichenketten belegt.

Um nun eine konkrete Adresse zu speichern, erzeugst du eine **Instanz** der Klasse (ein **Objekt**). Die Klammern hinter dem Klassennamen erzeugen die Instanz:

````python
chaosdorf = Adresse()   # die Klammern erzeugen eine Instanz (ein Objekt)

# Jetzt können wir die Eigenschaften des Objektes setzen:
chaosdorf.name = "Chaosdorf"
chaosdorf.strasse = "Hüttenstraße 25"
chaosdorf.plz = "40215"
chaosdorf.ort = "Düsseldorf"

print(chaosdorf.name)
print(chaosdorf.strasse)
print(chaosdorf.plz, chaosdorf.ort)
````

Eigenschaften, die wir gar nicht setzen (z.B. `vorname` oder `email`), behalten ihren Standardwert - die leere Zeichenkette aus der Klassendefinition.

Jedes Objekt hat außerdem seine **eigenen** Werte. Legen wir eine zweite Adresse an, hat diese ihre eigenen Eigenschaften, ohne die erste zu beeinflussen:

````python
freund = Adresse()
freund.name = "Mustermann"

print("Chaosdorf:", chaosdorf.name)
print("Freund:  ", freund.name)
````

## Bündelung von Daten und Methoden

Die Eigenschaften nach dem Erzeugen einzeln zu setzen, ist mühsam und fehleranfällig. Besser ist es, die Werte gleich **beim Erzeugen** mitzugeben. Dafür gibt es eine besondere Methode: den **Konstruktor** `__init__`.

Und: Funktionen, die zu einem Objekt gehören und auf dessen Daten zugreifen, nennt man **Methoden**. Ihr erster Parameter ist immer `self` - das ist eine Referenz auf das Objekt selbst.

````python
class Adresse:
    def __init__(self, name, strasse, plz, ort):
        # __init__ wird automatisch aufgerufen, wenn ein Objekt erzeugt wird.
        # Die Parameter werden als Eigenschaften (Attribute) gespeichert.
        self.name = name
        self.strasse = strasse
        self.plz = plz
        self.ort = ort

    def visitenkarte(self):
        """Gibt die Adresse hübsch formatiert aus."""
        print(self.name)
        print(self.strasse)
        print(f"{self.plz} {self.ort}")

    def __repr__(self):
        """Eine für Menschen lesbare Darstellung des Objekts."""
        return f"Adresse({self.name})"


# Beim Erzeugen geben wir die Werte direkt mit:
chaosdorf = Adresse("Chaosdorf", "Hüttenstraße 25", "40215", "Düsseldorf")
chaosdorf.visitenkarte()
````

Beim Aufruf `Adresse("Chaosdorf", ...)` ruft Python automatisch `__init__` auf. Die Methode `visitenkarte` rufst du über das Objekt auf: `chaosdorf.visitenkarte()`. Das `self` musst du dabei **nicht** mitgeben - Python setzt dort automatisch das Objekt ein, über das die Methode aufgerufen wird.

Die Methode `__repr__` ist ebenfalls besonders: Python benutzt sie, um ein Objekt lesbar darzustellen - etwa bei `print()` oder wenn das Objekt in einer Liste steckt. (Eng verwandt ist `__str__`.)

````python
print(chaosdorf)   # benutzt __repr__

# Das Pythonfoo trifft sich im Chaosdorf - gleiche Adresse, eigenes Objekt:
adressbuch = [chaosdorf, Adresse("Pythonfoo", "Hüttenstraße 25", "40215", "Düsseldorf")]
print(adressbuch)
````

### Klassen- und Instanzvariablen

In Abschnitt 1 haben wir die Eigenschaften als **Klassenvariablen** geschrieben (direkt im Klassenkörper), in Abschnitt 2 dagegen in `__init__` mit `self.…` als **Instanzvariablen**. Der Unterschied ist wichtig: Eine Klassenvariable wird von *allen* Instanzen geteilt.

Bei unveränderlichen Werten wie Zahlen oder Strings fällt das kaum auf. Bei *veränderlichen* Werten wie einer Liste wird es aber zur Falle:

````python
class Konto:
    buchungen = []           # ACHTUNG: Klassenvariable - von ALLEN Konten geteilt!

    def __init__(self, inhaber):
        self.inhaber = inhaber


anna = Konto("Anna")
ben = Konto("Ben")
anna.buchungen.append(50)    # nur bei Anna buchen ...

print("Annas Buchungen:", anna.buchungen)
print("Bens Buchungen: ", ben.buchungen)   # ... taucht auch bei Ben auf!
````

Beide Konten teilen sich dieselbe Liste! Deshalb legt man veränderliche Werte in `__init__` an - dort bekommt jede Instanz ihr **eigenes** Exemplar:

````python
class Konto:
    def __init__(self, inhaber):
        self.inhaber = inhaber
        self.buchungen = []   # jede Instanz bekommt ihre eigene Liste


anna = Konto("Anna")
ben = Konto("Ben")
anna.buchungen.append(50)

print("Annas Buchungen:", anna.buchungen)
print("Bens Buchungen: ", ben.buchungen)   # jetzt leer - korrekt
````

(Genau aus diesem Grund weisen die Klassen im Hogwarts-Quiz ihre Listen im Konstruktor mit `self.fragen = fragen` zu, obwohl oben im Klassenkörper schon `fragen = []` steht.)

### Ausblick: weniger Tipparbeit mit `dataclasses`

Für Klassen, die vor allem Daten bündeln (wie unsere `Adresse`), schreibt man `__init__` und `__repr__` oft fast gleich. Seit Python 3.7 nimmt einem die `@dataclass` diese Arbeit ab: Aus den aufgelisteten Feldern erzeugt sie automatisch `__init__`, `__repr__` und sogar `__eq__`.

````python
from dataclasses import dataclass


@dataclass
class Adresse:
    name: str
    strasse: str
    plz: str
    ort: str


a = Adresse("Chaosdorf", "Hüttenstraße 25", "40215", "Düsseldorf")
print(a)                       # __repr__ ist schon da
print(a == Adresse("Chaosdorf", "Hüttenstraße 25", "40215", "Düsseldorf"))  # __eq__ auch
````

## Vererbung

Eine Klasse muss nicht bei Null anfangen - sie kann auf einer anderen Klasse **aufbauen** und deren Eigenschaften und Methoden **erben**. Meistens erbt man von einer Klasse, die man selbst geschrieben hat.

Als Beispiel: Alle Tiere haben einen Namen und können ein Geräusch machen. Ein Hund und eine Katze sind spezielle Tiere - sie *erben* vom allgemeinen `Tier`, machen aber ihr eigenes Geräusch.

````python
class Tier:
    def __init__(self, name):
        self.name = name

    def sprich(self):
        print(f"{self.name} macht irgendein Geräusch.")


class Hund(Tier):              # Hund erbt von Tier - "ist ein" Tier
    def sprich(self):          # überschreibt die Methode von Tier
        print(f"{self.name} sagt: Wuff!")


class Katze(Tier):
    def __init__(self, name, farbe):
        super().__init__(name)   # den Konstruktor von Tier aufrufen
        self.farbe = farbe       # und ein eigenes Attribut ergänzen

    def sprich(self):
        print(f"{self.name} ({self.farbe}) sagt: Miau!")
````

````python
tiere = [Hund("Rex"), Katze("Minka", "getigert"), Tier("Etwas")]
for tier in tiere:
    tier.sprich()
````

`Hund` und `Katze` erben von `Tier` und überschreiben die Methode `sprich`. `Katze` erweitert zusätzlich den Konstruktor und ruft mit `super().__init__(name)` den der Basisklasse auf, damit der Name wie gehabt gesetzt wird. Im Loop ruft jedes Objekt *seine eigene* Version von `sprich` auf, obwohl der Code gleich aussieht - das nennt man **Polymorphie**.

Man kann sogar von **eingebauten** Typen wie `list` erben. Das nutzen wir für zwei klassische Beispiele: einen Stapel und eine Schlange.

Ein **Stapel** (englisch *stack*) funktioniert wie ein Stapel Teller: Du kannst immer nur den obersten (zuletzt abgelegten) nehmen - *last in, first out*. Eine Liste bietet schon alles Nötige, deshalb ist die Klasse denkbar einfach:

````python
class Stack(list):          # Stack ("Stapel") erbt von list
    def is_empty(self):
        """gibt True zurück, wenn der Stapel leer ist"""
        return len(self) == 0

    def push(self, element):
        """legt ein Element oben auf den Stapel"""
        self.append(element)    # append() ist von list geerbt

    def pop(self):
        """holt das oberste (zuletzt abgelegte) Element"""
        return super().pop()    # ruft die pop()-Methode der Liste auf
````

````python
stapel = Stack()
stapel.push("Teller 1")
stapel.push("Teller 2")
stapel.push("Teller 3")
print("Stapel:    ", stapel)
print("oben weg:  ", stapel.pop())
print("oben weg:  ", stapel.pop())
print("Stapel jetzt:", stapel)
print("leer?      ", stapel.is_empty())
````

Die Methode `pop()` gibt es bei Listen eigentlich schon - wir müssten sie gar nicht definieren. Wir tun es hier trotzdem, denn *explizit ist besser als implizit*.

Eine **Schlange** (englisch *queue*) funktioniert umgekehrt: Wer sich zuerst anstellt, kommt zuerst dran - *first in, first out*, wie an der Supermarktkasse. Wir holen also nicht das letzte, sondern das **erste** Element. Dafür gibt es `pop(0)`:

````python
class Queue(list):          # Queue = "Schlange"
    def is_empty(self):
        return len(self) == 0

    def push(self, element):
        """hängt ein Element hinten an die Schlange an"""
        self.append(element)

    def pop(self):
        """holt das erste (zuerst angestellte) Element"""
        return super().pop(0)   # pop(0) holt und entfernt das erste Element


schlange = Queue()
schlange.push("Dokument A")
schlange.push("Dokument B")
schlange.push("Dokument C")
print("druckt zuerst:", schlange.pop())
print("dann:         ", schlange.pop())
````

Ein typischer Anwendungsfall ist die **Druckerschlange**: Dein Computer legt zu druckende Dokumente in einer Schlange ab, und der Drucker holt sich immer das nächste. So kommt das zuerst abgeschickte Dokument auch zuerst aus dem Drucker.

`pop(0)` funktioniert zwar, aber die Standardbibliothek hat einen Typ, der dafür noch besser geeignet ist: `deque` (für *doubly ended queue*, "Schlange mit zwei Enden"). Er kann Elemente am Anfang schneller entfernen als eine Liste. `deque` ist nicht eingebaut, sondern muss importiert werden. Neu sind die Methoden `appendleft()` und `popleft()`, die am **Anfang** arbeiten:

````python
from collections import deque

class Queue(deque):         # jetzt auf Basis von deque statt list
    def is_empty(self):
        return len(self) == 0

    def push(self, element):
        self.append(element)

    def pop(self):
        """holt das erste Element - popleft() ist schneller als pop(0)"""
        return super().popleft()


schlange = Queue()
schlange.push("A")
schlange.push("B")
schlange.push("C")
print(schlange.pop(), schlange.pop(), schlange.pop())
````

## Operationen auf Objekten

Bisher haben wir Methoden mit einem Punkt aufgerufen (`chaosdorf.visitenkarte()`). Wir können aber auch die **Rechenoperatoren** `+`, `-`, `*`, `/` und `**` für eigene Objekte definieren. Python ruft dafür im Hintergrund spezielle Methoden auf. Das probieren wir an der **Bruchrechnung** aus.

Ein Bruch besteht aus zwei ganzen Zahlen: dem **Zähler** (oben) und dem **Nenner** (unten). Den Wert eines Bruchs erhältst du, wenn du den Zähler durch den Nenner teilst. Wir treffen die Übereinkunft, dass der Nenner immer positiv ist - ein negativer Bruch hat also einen negativen Zähler.

Wir schreiben den Kern der Klasse: Konstruktor, lesbare Darstellung, Kürzen (mit dem Euklidischen Algorithmus für den größten gemeinsamen Teiler), Addition und einen Vergleich.

````python
class Bruch:
    def __init__(self, zähler, nenner):
        if nenner < 0:                       # Vorzeichen immer an den Zähler
            zähler, nenner = -zähler, -nenner
        if nenner == 0:
            raise ZeroDivisionError("Nenner eines Bruchs darf nicht Null sein")
        self.zähler = zähler
        self.nenner = nenner
        self.kürzen()

    def __repr__(self):
        return f"({self.zähler}/{self.nenner})"

    @staticmethod
    def ggt(a, b):
        """größter gemeinsamer Teiler nach dem Euklidischen Algorithmus"""
        while b != 0:
            a, b = b, a % b
        return a

    def kürzen(self):
        g = Bruch.ggt(self.zähler, self.nenner)
        self.zähler //= g
        self.nenner //= g

    def __add__(self, rechts):               # wird aufgerufen bei  Bruch + x
        if isinstance(rechts, int):
            rechts = Bruch(rechts, 1)        # eine ganze Zahl wird zum Bruch
        return Bruch(self.zähler * rechts.nenner + self.nenner * rechts.zähler,
                     self.nenner * rechts.nenner)

    def __radd__(self, links):               # wird aufgerufen bei  x + Bruch
        return self + links

    def __eq__(self, anderer):               # wird aufgerufen bei  Bruch == x
        if isinstance(anderer, int):
            anderer = Bruch(anderer, 1)
        if not isinstance(anderer, Bruch):
            return NotImplemented
        # Beide Brüche sind immer gekürzt, daher genügt der direkte Vergleich:
        return self.zähler == anderer.zähler and self.nenner == anderer.nenner

    # (Wer __eq__ definiert, sollte auch __hash__ definieren - siehe
    #  Lösung zu Aufgabe 9.3.)
````

````python
print("1/2 + 1/3 =", Bruch(1, 2) + Bruch(1, 3))
print("1/2 + 1   =", Bruch(1, 2) + 1)    # __add__ wandelt die 1 in einen Bruch
print("1 + 1/2   =", 1 + Bruch(1, 2))    # hier springt __radd__ ein
print("6/8 wird automatisch gekürzt:", Bruch(6, 8))
print("1/2 == 2/4 ?", Bruch(1, 2) == Bruch(2, 4))   # __eq__

# Mit assert kannst du dein Programm selbst prüfen - läuft es ohne Fehler durch,
# stimmt die Bedingung:
assert Bruch(6, 8) == Bruch(3, 4)
print("Selbsttest bestanden.")
````

Zwei Dinge sind hier neu:

* **`isinstance(rechts, int)`** prüft, ob ein Objekt von einem bestimmten Typ ist (hier: eine ganze Zahl). So kann `__add__` eine ganze Zahl zuerst in einen Bruch umwandeln. Eng verwandt, aber verschieden ist das Schlüsselwort `is`: Es prüft, ob zwei Namen auf *dasselbe* Objekt zeigen (Identität), während `==` auf *gleichen Wert* prüft.
* **`__eq__`** definiert, wann zwei Objekte als gleich gelten. Weil im Konstruktor immer gekürzt wird, sind `Bruch(1, 2)` und `Bruch(2, 4)` tatsächlich gleich.

Bei jeder Rechnung wird ein **neues** `Bruch`-Objekt erzeugt und zurückgegeben - und im Konstruktor gleich gekürzt. Die übrigen Operatoren funktionieren analog:

| Rechenoperation | Python-Methode |
|---|---|
| Addition (`+`) | `__add__`, `__radd__` |
| Negation (unäres Minus) | `__neg__` |
| Subtraktion (`-`) | `__sub__`, `__rsub__` |
| Multiplikation (`*`) | `__mul__`, `__rmul__` |
| Division (`/`) | `__truediv__`, `__rtruediv__` |
| Potenz (`**`) | `__pow__` |

## Rückblick

In diesem Level hast du die Grundlagen der objektorientierten Programmierung kennengelernt:

* Eine **Klasse** fasst **Daten** (Attribute) und **Verhalten** (Methoden) zu einer Einheit zusammen. Aus einer Klasse erzeugst du **Objekte** (Instanzen).
* Der Konstruktor **`__init__`** legt die Attribute beim Erzeugen fest; der erste Parameter jeder Methode ist **`self`**.
* **Klassenvariablen** werden von allen Instanzen geteilt, **Instanzvariablen** (in `__init__` mit `self.…`) gehören zu einem einzelnen Objekt - wichtig besonders bei veränderlichen Werten wie Listen.
* Besondere Methoden wie **`__repr__`** / **`__str__`** bestimmen die Darstellung, **`__eq__`** den Vergleich.
* Mit **Vererbung** baut eine Klasse auf einer anderen auf (eigener oder eingebauter) und greift mit **`super()`** auf deren Methoden zu; überschriebene Methoden führen zu **Polymorphie**.
* Durch **Operator-Methoden** (`__add__`, `__mul__`, ...) kannst du Rechenoperatoren für eigene Objekte definieren.
* `@dataclass` erzeugt für reine Daten-Klassen `__init__`, `__repr__` und `__eq__` automatisch.
